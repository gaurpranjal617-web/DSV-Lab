```python
s = input("Enter a String")
count = 0

for ch in s.lower():
  if ch in "aeiou":
    count += 1

print("Vowels:", count)
```

```python
n = int(input("Enter a number: "))
fact = 1

for i in range(1, n+1):
  fact *= i

print("Factorial:", fact)
```

```python
n = int(input("Enter a number: "))
n = abs(n)
total = 0

while n > 0:
  digit = n%10
  total += digit
  n //=10

print("Sum of digits: ", total)
```
